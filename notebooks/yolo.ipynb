```python
from google.colab import drive
drive.mount('/content/drive')

import os, cv2
os.chdir("/content/drive/MyDrive/Vehicle-type-classification")


from src.detector import VehicleDetector
from src.utils import draw_detection
```

```python
detector = VehicleDetector(
    model_path="yolov8n.pt",
    conf=0.4
)
```

```python
video_path = "data/test-video/1900-151662242_medium.mp4"

output_path = "outputs/detected_video.mp4"

os.makedirs("outputs", exist_ok=True)
```

```python
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print("Error opening video")

width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = int(cap.get(cv2.CAP_PROP_FPS))

fourcc = cv2.VideoWriter_fourcc(*"mp4v")

out = cv2.VideoWriter(
    output_path,
    fourcc,
    fps,
    (width, height)
)

while True:

    ret, frame = cap.read()

    if not ret:
        break

    detections = detector.detect(frame)

    for detection in detections:

        frame = draw_detection(
            frame,
            detection
        )

    out.write(frame)

cap.release()
out.release()

print("Video saved:", output_path)
```
