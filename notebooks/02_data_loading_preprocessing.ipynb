```python
from google.colab import drive
drive.mount('/content/drive')

import os, json
os.chdir("/content/drive/MyDrive/Vehicle-type-classification")
```

```python
DATASET_DIR = "data/classification-dataset"

config = {
    "dataset_dir": DATASET_DIR,
    "image_height": 224,
    "image_width": 224,
    "batch_size": 32,
    "validation_split": 0.2
}
```

```python
class_names = sorted([
    d for d in os.listdir(DATASET_DIR)
    if os.path.isdir(os.path.join(DATASET_DIR, d))
])

class_indices = {class_name: idx for idx, class_name in enumerate(class_names)}

print("Detected classes:", class_indices)
```

```python
os.makedirs("artifacts", exist_ok=True)

with open("artifacts/config.json", "w") as f:
    json.dump(config, f)

print("Config saved")
```

```python
with open("artifacts/class_indices.json", "w") as f:
    json.dump(class_indices, f)

print("Class indices saved successfully")
```

```python
import json

with open("artifacts/config.json", "r") as f:
    loaded_config = json.load(f)
print("Config (artifacts/config.json):")
display(loaded_config)

with open("artifacts/class_indices.json", "r") as f:
    loaded_class_indices = json.load(f)
print("\nClass Indices (artifacts/class_indices.json):")
display(loaded_class_indices)
```
