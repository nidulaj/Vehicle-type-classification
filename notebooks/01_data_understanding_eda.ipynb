Purpose:

- Verify class folders and image availability

- Identify class imbalance issues

- Inspect sample images for label correctness

- Check image size variability and data quality issues

- Document data assumptions before preprocessing and modeling

```python
import os
import cv2
import matplotlib.pyplot as plt
from google.colab import drive
drive.mount('/content/drive')

import os
os.chdir("/content/drive/MyDrive/Vehicle-type-classification")
```

```python
dataset_path = "data/classification-dataset"

class_counts = {}
for class_name in os.listdir(dataset_path):
  class_path = os.path.join(dataset_path, class_name)
  if os.path.isdir(class_path):
    class_counts[class_name] = len(os.listdir(class_path))

class_counts
```

```python
plt.bar(class_counts.keys(), class_counts.values())
plt.title("Number of images per class")
plt.xlabel("Class")
plt.ylabel("Number of images")
```

```python
import random

plt.figure(figsize=(10,6))
for i, class_name in enumerate(class_counts.keys()):
  img_name = random.choice(os.listdir(os.path.join(dataset_path, class_name)))
  img_path = os.path.join(dataset_path, class_name, img_name)
  print(img_path)

  img = cv2.imread(img_path)
  img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

  plt.subplot(3,2,i+1)
  plt.imshow(img)
  plt.title(class_name)
  plt.axis("off")

plt.show()
```

```python
sizes = []

for class_name in os.listdir(dataset_path):
  class_path = os.path.join(dataset_path, class_name)
  for img_name in os.listdir(class_path):
    img = cv2.imread(os.path.join(class_path, img_name))
    if img is not None:
      sizes.append(img.shape[:2])

sizes[:5]
```
